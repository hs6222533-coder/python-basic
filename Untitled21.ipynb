```python
# Q7. Convert Order_Date into datetime format. Find the earliest and latest order date.
df["Order_Date"] = pd.to_datetime(df["Order_Date"], errors="coerce")

df["Order_Date"].agg(['max','min'])
```

```python
# Q8. Find all orders where Quantity is greater than 3, Rating is less than or equal to 2, and Order_Status is Delivered.
df[(df["Quantity"] > 3) & (df["Rating"] <= 2) & (df["Order_Status"] == "Delivered")]
```

```python
# Q9. Find the top 10 orders based on Sales. Display Order_ID, Customer_Name, Product, Quantity, and Sales.
df.sort_values("Sales", ascending=False).head(10)[["Order_ID", "Customer_Name", "Product", "Quantity", "Sales"]]
```

```python
# Q10. Find all orders where Discount is greater than 20 and sort them from highest to lowest Discount.
df[df["Discount"] > 0.20].sort_values("Discount", ascending=False)
df
```

```python
# Q11. Find the 10 products with the highest average rating. Display the product name and average rating.
df.groupby("Product")["Rating"].mean().sort_values(ascending=False).head(10)
```

```python
# Q12. Calculate the total sales generated by each Category. Sort the result from highest to lowest sales.
df.groupby("Category", as_index=False)["Sales"].sum().sort_values("Sales", ascending=False)
```

```python
# Q13. Find the total quantity sold and total sales for each Product. Display the top 10 products based on total sales.
df.groupby("Product", as_index=False).agg({
    "Quantity": "sum",
    "Sales": "sum"
}).sort_values("Sales", ascending=False).head(10)
```

```python
# Q14. Calculate the average Sales, average Rating, and average Delivery_Days for each Payment_Mode.
df.groupby("Payment_Mode", as_index=False).agg({
    "Sales": "mean",
    "Rating": "mean",
    "Delivery_Days": "mean"
})
```

```python
# Q15. Find the number of orders and total sales generated by each City. Which five cities have generated the highest sales?
df.groupby("City", as_index=False).agg(
    Orders=("Order_ID", "count"),
    Total_Sales=("Sales", "sum")
).sort_values("Total_Sales", ascending=False).head(5)
```

```python
# Q16. Create a new column called Month from Order_Date. Find the total sales for each month.
df["Order_Date"] = pd.to_datetime(df["Order_Date"], errors="coerce")

df["Month"] = df["Order_Date"].dt.month_name()

df.groupby("Month")["Sales"].sum()
```

```python
# Q17. Find the total sales for each Order_Status. Also calculate what percentage of total orders belong to each status.
result = df.groupby("Order_Status").agg(
    Total_Sales=("Sales", "sum"),
    Total_Orders=("Order_Status", "count"))

result["Order_Percentage"] = (result["Total_Orders"] / len(df) * 100)

result
```

```python
# Q18. Compare the performance of different categories by calculating Total Sales, Average Sales, Total Quantity Sold, Average Rating, and Average Delivery Days. Which category appears to perform best overall?
df.groupby("Category").agg(
    Total_Sales=("Sales", "sum"),
    Average_Sales=("Sales", "mean"),
    Total_Quantity=("Quantity", "sum"),
    Average_Rating=("Rating", "mean"),
    Average_Delivery_Days=("Delivery_Days", "mean"))
```

```python
# Q19. Create a new column called Calculated_Sales using: Quantity × Unit_Price × (1 − Discount). Compare it with the existing Sales column and find the rows where the difference is greater than 1.
df["Calculated_Sales"] = df["Quantity"] * df["Unit_Price"] * (1 - df["Discount"])

df["Difference"] = abs(df["Calculated_Sales"] - df["Sales"])

df[df["Difference"] > 1]
```

```python
# Q20. Create a new column called Customer_Segment based on Customer_Age: Age < 25 = Young; 25–40 = Adult; 41–60 = Middle Age; Above 60 = Senior. Then calculate the number of orders, total sales, and average rating for each customer segment.
def customer_segment(age):
    if age < 25:
        return "Young"
    elif age <= 40:
        return "Adult"
    elif age <= 60:
        return "Middle Age"
    else:
        return "Senior"

df["Customer_Segment"] = df["Customer_Age"].apply(customer_segment)

df.groupby("Customer_Segment").agg(
    Total_Orders=("Order_ID", "count"),
    Total_Sales=("Sales", "sum"),
    Average_Rating=("Rating", "mean")
)
```

```python
# Final Business Case – 5 Business Insights
# 1. Category Performance
category_analysis = df.groupby("Category").agg(
    Total_Sales=("Sales", "sum")
    Average_Sales=("Sales", "mean"),
    Total_Quantity=("Quantity", "sum"),
    Average_Rating=("Rating", "mean"),
    Average_Delivery_Days=("Delivery_Days", "mean")
).sort_values("Total_Sales", ascending=False)

category_analysis
```

```python
# 2. Discount Analysis
discount_analysis = df.groupby("Discount").agg(
    Total_Sales=("Sales", "sum"),
    Average_Sales=("Sales", "mean"),
    Total_Orders=("Sales", "count")
).sort_values("Total_Sales", ascending=False)

discount_analysis
```

```python
# 3. Rating Analysis
df.groupby("Rating").agg(
    Total_Sales=("Sales", "sum"),
    Average_Sales=("Sales", "mean"),
    Total_Orders=("Sales", "count")
).sort_values("Rating", ascending=False)
```

```python
# 4. Delivery Analysis
df.groupby("Delivery_Days").agg(
    Total_Sales=("Sales", "sum"),
    Average_Sales=("Sales", "mean"),
    Total_Orders=("Sales", "count")
).sort_values("Delivery_Days")
```

```python
# 5. Customer Segment Analysis
df.groupby("Customer_Segment").agg(
    Total_Orders=("Order_ID", "count"),
    Total_Sales=("Sales", "sum"),
    Average_Sales=("Sales", "mean"),
    Average_Rating=("Rating", "mean")
).sort_values("Total_Sales", ascending=False)
```
