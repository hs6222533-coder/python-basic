```python
df = pd.DataFrame({
    "Product" : ["laptop","notebook","book","mouse","book stand","keyboard"],
    "City" : ["jaipur","delhi","mumbai","jaipur","delhi","pune"],
    "Sales" : [50000,8000,6000,3000,12000,15000],
    "Rating" : [4.5,4.0,4.2,3.8,4.5,4.1]
},index = [101,102,103,104,105,106])

print(df)
# df = pd.DataFrame(df)
df_1 = df[df['Sales'] > 8000]
print(df_1)
```

```python
print(df[df['City']=='delhi'])
```
